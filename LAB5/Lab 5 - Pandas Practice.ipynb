### **Part 3: Data Grouping and Summary**

Q1: Calculate the sum of all values ​​in column a.

Q2: Calculate the average of column b.

Q3: Find the largest and smallest values ​​in column c.

Q4: Use `.describe()` to print the general statistics of the numerical columns in df.

Q5: Group the DataFrame by column d and calculate the sum of the values ​​in each group using `.groupby()`.

Q6: Group by column d and count the number of rows in each group.

Q7: Group by column d and calculate the average of column a.

Q8: Calculate the standard deviation (std) of column b.

Q9: Calculate the variance (var) of column a.

Q10: Calculate the cumulative sum (sum from top to bottom) of column b using `.cumsum()`.

```python
# Q1: Calculate the sum of all values in column a
sum_a = df['a'].sum()
print("Q1:", sum_a)


# Q2: Calculate the average of column b
avg_b = df['b'].mean()
print("Q2:", avg_b)


# Q3: Find the largest and smallest values in column c
max_c = df['c'].max()
min_c = df['c'].min()
print("Q3 - Largest:", max_c)
print("Q3 - Smallest:", min_c)


# Q4: Use `.describe()` to print the general statistics of the numerical columns in df.
summary = df.describe()
print("Q4:")
print(summary)


# Q5: Group the DataFrame by column d and calculate the sum of the values ​​in each group using `.groupby()`.
group_sum = df.groupby('d').sum()
print("Q5:")
print(group_sum)


# Q6: Group by column d and count the number of rows in each group.
group_count = df.groupby('d').size()
print("Q6:")
print(group_count)


# Q7: Group by column d and calculate the average of column a.
group_avg = df.groupby('d')['a'].mean()
print("Q7:")
print(group_avg)


# Q8: Calculate the standard deviation (std) of column b.
std_b = df['b'].std()
print("Q8:", std_b)


# Q9: Calculate the variance (var) of column a.
var_a = df['a'].var()
print("Q9:", var_a)


# Q10: Calculate the cumulative sum (sum from top to bottom) of column b using `.cumsum()`.
cumulative_sum = df['b'].cumsum()
print("Q10:")
print(cumulative_sum)
```

### **Part 4: Combining and Handling Missing Data**

Q1: Check if the DataFrame contains any missing values ​​(NaN) using `isnull().any()`.

Q2: Count the number of missing values ​​(NaN) in each column.

Q5: Completely delete the rows containing NaN values ​​using `.dropna()`.

Q6: Subtract the values ​​of column c from column b and store them in a new column named `hieu`.

Q7: Calculate the mathematical expression: (a^2 + b^2) and store it in the `tinh_toan` column.

Q8: Create `df_them` = `pd.DataFrame({"a": [10], "b": [10], "c": [10], "d": ['W']})`. Concatenate df_them vertically to the bottom of df using pd.concat().

Q9: Concatenate pd.concat([df, df_them], axis=1) horizontally and observe the result.

Q10: Use .clip(lower=5, upper=12) on column c to change values ​​less than 5 to 5 and greater than 12 to 12.

```python
# Q1: Check if the DataFrame contains any missing values ​​(NaN) using isnull().any()
has_nan = df.isnull().any()
print("Q1")
print(has_nan)


# Q2: Count the number of missing values ​​(NaN) in each column.
nan_count = df.isnull().sum()
print("Q2")
print(nan_count)


# Q5: Completely delete the rows containing NaN values ​​using .dropna().
df_no_nan = df.dropna()
print("Q5")
print(df_no_nan)


# Q6: Subtract the values ​​of column c from column b and store them in a new column named hieu.
df['hieu'] = df['b'] - df['c']
print("Q6")
print(df)


# Q7: Calculate the mathematical expression: (a^2 + b^2) and store it in the tinh_toan column.
df['tinh_toan'] = df['a']**2 + df['b']**2
print("Q7")
print(df)


# Q8: Create df_them = pd.DataFrame({"a": [10], "b": [10], "c": [10], "d": ['W']}). 
# Concatenate df_them vertically to the bottom of df using pd.concat().
df_them = pd.DataFrame({
    "a": [10],
    "b": [10],
    "c": [10],
    "d": ['W']
})

df_concat_vertical = pd.concat([df, df_them], ignore_index=True)
print("Q8")
print(df_concat_vertical)


# Q9: Concatenate pd.concat([df, df_them], axis=1) horizontally and observe the result.
df_concat_horizontal = pd.concat([df, df_them], axis=1)
print("Q9")
print(df_concat_horizontal)


# Q10: Use .clip(lower=5, upper=12) on column c to change values ​​less than 5 to 5 and greater than 12 to 12.
df['c_clipped'] = df['c'].clip(lower=5, upper=12)
print("Q10")
print(df[['c', 'c_clipped']])
```

### **Part 5: Data Types and Strings**

Q1. Check the data type of each column in df using the `.dtypes` attribute.

Q2. Convert column a from integer (int) to floating-point (float) using `.astype(float)`.

Q3. Convert column b to a string.

Q4. Use `.str.lower()` to convert the letters in column d to lowercase.

Q5. Replace the character 'X' in column d with the character 'A' using `.str.replace()`.

Q6. Check if column d contains the character 'Y' using `.str.contains()`.

Q7. Add the word "Group" after each value in column d (e.g., 'X' becomes 'X Group').

Q8. Create column d_lap by copying the string in column d (e.g., 'X' becomes 'XX').

Q9. Use the `np.sqrt()` function in Numpy to calculate the square root of all values ​​in column a.

Q10. Use the `np.log()` function to calculate the natural logarithm of the values ​​in column b.

```python
# Q1. Check the data type of each column in df using the .dtypes attribute.
print("Q1")
print(df.dtypes)


# Q2. Convert column a from integer (int) to floating-point (float) using .astype(float).
df['a'] = df['a'].astype(float)
print("Q2")
print(df['a'])


# Q3. Convert column b to a string.
df['b'] = df['b'].astype(str)
print("Q3")
print(df['b'])


# Q4. Use .str.lower() to convert the letters in column d to lowercase.
df['d_lower'] = df['d'].str.lower()
print("Q4")
print(df['d_lower'])


# Q5. Replace the character 'X' in column d with the character 'A' using .str.replace().
df['d_replace'] = df['d'].str.replace('X', 'A')
print("Q5")
print(df['d_replace'])


# Q6. Check if column d contains the character 'Y' using .str.contains().
contains_y = df['d'].str.contains('Y')
print("Q6")
print(contains_y)


# Q7. Add the word "Group" after each value in column d (e.g., 'X' becomes 'X Group').
df['d_group'] = df['d'] + ' Group'
print("Q7")
print(df['d_group'])


# Q8. Create column d_lap by copying the string in column d (e.g., 'X' becomes 'XX').
df['d_lap'] = df['d'] * 2
print("Q8")
print(df['d_lap'])


# Q9. Use the np.sqrt() function in Numpy to calculate the square root of all values ​​in column a.
df['sqrt_a'] = np.sqrt(df['a'])
print("Q9")
print(df['sqrt_a'])


# Q10. Use the np.log() function to calculate the natural logarithm of the values ​​in column b.
df['b'] = df['b'].astype(float)  # convert back to numeric for log
df['log_b'] = np.log(df['b'])
print("Q10")
print(df['log_b'])
```
