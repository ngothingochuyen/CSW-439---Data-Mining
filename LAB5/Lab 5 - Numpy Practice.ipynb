### **Q9: Dot Product of Two Matrices**

Write a function that takes two matrices `X` and `W` as input and returns the dot product `M` using the following formula:

```
M = (x^T)W
```
Where `x^T` is the transpose of `x`. The dimension needs to be checked to ensure the calculation is always possible before performing the calculation, returning `None` if the dimension is incompatible.

Example:

Ví dụ:


Input:
```
X = np.array([
    [1, 1, 1],
    [0, 0, 0]]
)

W = np.array([
    [1, 1, 1],
    [0, 0, 0]]
)
```

Output:
```
[[1, 1, 1],
 [1, 1, 1],
 [1, 1, 1]]
```

````python
def product_vec_mat(X: np.ndarray, W: np.ndarray) -> Optional[np.ndarray]:
    ### BEGIN SOLUTION
    XT = X.T
    if XT.shape[1] != W.shape[0]:
        return None
    return np.dot(XT, W)
    ### END SOLUTION
````

````python
# Sample Input & Output
X = np.array([
  [1, 1, 1],
  [0, 0, 0]
])
W = np.array([
  [1, 1, 1],
  [0, 0, 0]
])
product_vec_mat(X, W)
````

### **Q11: Savings Deposit**

Calculate the remaining balance in a bank account with the input being a list of transaction amounts, corresponding to a list of deposit transaction types (debit, denoted by the string 'D'), withdrawal transaction types (debit, denoted by the string 'W')

**Input:**
- moneys: a numpy array containing a list of transaction amounts (deposits and withdrawals from the account)

- types: a numpy array containing a list of transaction types (deposits denoted by 'D', withdrawals denoted by 'W')

**Output:**
- A real number, the remaining balance in the account

Example:

Input:
```
moneys = [400.0, 100.0, 200.0]
types = ['D', 'W', 'D']
```

Output:
```
500.0
```

````python
def balance(moneys: np.ndarray, types: np.ndarray) -> float:
    ### BEGIN SOLUTION
    moneys = np.array(moneys, dtype=float)
    types = np.array(types)

    deposits = np.sum(moneys[types == 'D'])
    withdrawals = np.sum(moneys[types == 'W'])

    return deposits - withdrawals
    ### END SOLUTION
````

````python
# Sample Input & Output
balance(np.array([400.0, 100.0, 200.0]), np.array(['D', 'W', 'D']))
````

````python
# Public Test
assert balance(np.array([400.0, 100.0, 200.0]), np.array(['D', 'W', 'D'])) == 500
````

### **Q12: Average Grade**

Calculate the average grade of a student given a list of grades and credits for each course.

**Input:**
- marks: List type, containing *n* elements. This is the grade for each course.
- credits: List type, containing *n* elements. This is the number of credits for each course.

**Output:**
- Real number type, the average grade for each course.

Example:

Input:
```
marks = [8.0, 9.0, 10.0]
credits = [2, 2, 1]
```

Output:
```
8.8
```

````python
def gpa_one_student(marks: List[float], credits: List[int]) -> float:
    result = 0
    ### BEGIN SOLUTION
    marks = np.array(marks, dtype=float)
    credits = np.array(credits)

    # Calculate weighted average
    result = np.dot(marks, credits) / np.sum(credits)
    ### END SOLUTION
    return result
````

````python
# Sample Input & Output
marks = [8.0, 9.0, 10.0]
credits = [2, 2, 1]
gpa_one_student(marks, credits)
````

### **Q15**

Calculate the cumulative GPA of each student in a class given a list of scores and credits for each subject in the class.

Note: When calculating the GPA, we will exclude scores and credits for subjects below 5 points (failed).

**Formula for calculating the cumulative GPA:**
- GPA = (score for subject 1 * number of credits for subject 1 + ... + score for subject n * number of credits for subject n) / total number of credits

**Input:**
- marks: numpy array, size m x n. This is the score for each subject in the class.
    - m is the number of students
    - n is the number of subjects.
- credits: numpy array, size 1 x n. This is the number of credits for each subject.

**Output:**
- numpy array, size 1 x m, containing the average scores of the students in the class.

Example:

Input:
```
marks = np.array([
    [8.0, 9.0, 10.0],
    [4.0, 9.0, 8.0],
    [8.0, 3.0, 8.0],
    [10.0, 9.0, 5.0],
    [9.0, 9.0, 4.0]
])

credits = np.array([2, 2, 1])
```

Output:
```
array([8.8, 8.66666667, 8., 8.6, 9.])
```

````python
def gpa_of_pass(marks: np.ndarray, credits: np.array) -> np.ndarray:
    ### BEGIN SOLUTION
    passed = marks >= 5

    weighted_scores = np.sum(marks * credits * passed, axis=1)

    total_credits = np.sum(credits * passed, axis=1)

    return weighted_scores / total_credits
    ### END SOLUTION
````

````python
# Sample Input & Output
marks = np.array(
    [[8.0, 9.0, 10.0],
    [4.0, 9.0, 8.0],
    [8.0, 3.0, 8.0],
    [10.0, 9.0, 5.0],
    [9.0, 9.0, 4.0]])
credits = np.array([2,2,1])
gpa_of_pass(marks, credits)
````

````python
# Public Test
marks = np.array(
   [[4.0, 9.0, 4.5],
   [5.0, 8.0, 4.5],
   [6.0, 7.0, 7.5],
   [7.0, 6.0, 7.5],
   [8.0, 5.0, 4.5]])
credits = np.array([2,1,2])
assert np.sum(abs(gpa_of_pass(marks, credits)-np.array([9., 6., 6.8, 7.0, 7.0]))) < 0.000001
````

##### **Requirement**: Implement `Fraction` class.

````python
import math


class Fraction:
    def __init__(self, numerator: int, denominator: int) -> None:
        # Check denominator
        assert denominator != 0, "Denominator cannot be 0"

        # English attributes
        self.numerator = numerator
        self.denominator = denominator

        # Vietnamese aliases
        self.tu_so = self.numerator
        self.mau_so = self.denominator

        self.reduction()

    def reduction(self) -> None:
        if self.numerator == 0:
            self.denominator = 1
        else:
            gcd = math.gcd(self.numerator, self.denominator)

            self.numerator //= gcd
            self.denominator //= gcd

            # Make denominator positive
            if self.denominator < 0:
                self.numerator *= -1
                self.denominator *= -1

        # Update aliases
        self.tu_so = self.numerator
        self.mau_so = self.denominator

    def add(self, other: 'Fraction') -> None:
        self.numerator = (
            self.numerator * other.denominator
            + other.numerator * self.denominator
        )

        self.denominator *= other.denominator
        self.reduction()

    def subtract(self, other: 'Fraction') -> None:
        self.numerator = (
            self.numerator * other.denominator
            - other.numerator * self.denominator
        )

        self.denominator *= other.denominator
        self.reduction()

    # Vietnamese alias
    def tru(self, other: 'Fraction') -> None:
        self.subtract(other)
````
